```python
# Removing punctuations
import string
text = "Rahul!!! loves coding, and Python@@ is his favorite language."
p= word_tokenize(text)
print(p)

words= []
for i in p:
    if i not in string.punctuation:
        words.append(i)
print(words)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
#Stop word Removal
from nltk.corpus import stopwords
nltk.download('stopwords')

text="Ram is a good cook, but he can not make pulses."
p= word_tokenize(text)
words=[]
for i in p:
    if i.lower() not in stopwords.words('english'):
        words.append(i)
print(words)
```

```python
# Spelling checking 1
!pip install pyspellchecker
from spellchecker import SpellChecker
spell = SpellChecker()
text = "I have a good, bad, and ugly situation"
p = text.split()
correct = []
for i in p:
    correct.append(spell.correction(i))
print(correct)
```
